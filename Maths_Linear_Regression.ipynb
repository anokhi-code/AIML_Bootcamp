```python
# limnear regression
y = B0 + B1x1 + B2X2+...+BpXp +E

where
y= dependent variable(what we want to predict)
xi,x2,..xp= independent variable(features)
B0 = intercept(value of y when all x are 0)
B1, B2...,Bp = coefficients(weights for each feature)
e = error term( difference between predicted and actual value)

In simple linear regression (one feature):
y= B0+B1x+E

objective fuctioon-
goal is to find B0, B1,...Bp, that minimize the error, Mosst commonly we minimize the sum of squared error(SSE):
SSE=E(yi-y^i)2
y^i= B0,B1xi1+...BpXip is the predicted value
minimizing SSE is also called Ordinary Least Square(OLS)

Matrix Formulation(vectorized)
For multiple variables , its easier to write in matrix forms:
y=XB+E
y= N*1 vector of outputs
X= n*(p+1) matrix of inputs(first column is all 1s for B0)
B=(p+1)* 1 vesctor of coefficients
E= n* 1 vector of errors

the loss function becomes
L(B)=(y-XB)T(y-XB)

long functions measures how wrong a model's predictions are compared to actual value.


p is how many features you are using to predict y
predict house price with 3 features
1. size in sqft - x1
2. number of bedroom - x2
3. year of house- x3
p=3, total columns in X=P+1=4(including the intercept column)
B= [BO,B1,B2,B3]T
thre model multiplies X*B---gives predicted y for all data points
```
