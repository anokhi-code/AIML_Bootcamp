```python
# Use Case : -- Predicting Iris Flower Species
# Goal - Build a cdecision tree model to classify iris flower into species based on ttheie sepal and petal measurement
objective- train the model, visualize the decision rules, and make a prediction for a new flower
```

```python
#1. import library
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree
import matplotlib.pyplot as plt
```

```python
# 2. load iris dataset
iris = load_iris()
```

```python
#3. convert dataset into pandas Dataframe for better readability
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['target'] = iris.target
```

```python
#4.-split the datasetinto features(x) and target(y)
X = df[iris.feature_names]
y = df['target']
```

```python
# 5.- intialize decision tree classifier
model = DecisionTreeClassifier(random_state = 42)
```

```python
# 6 - train model using the dataset
model.fit(X,y)
```

```python
# 7- Visualize the Decision Tree
plt.figure(figsize=(15,10))
tree.plot_tree(model,
               feature_names=iris.feature_names,
               class_names=iris.target_names,
               filled=True,
               rounded=True)
plt.show()
```

```python
# 8 - make a prediction on a new flower data
new_flower = [[6.0,3.0,4.8,1.8]]
predicted_class = model.predict(new_flower)
predicted_species = iris.target_names[predicted_class][0]
print("prediction for new flower:", predicted_species)
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

```python
iris = load_iris()
X,y =iris.data, iris.target

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size= 0.3, random_state=42)

```

```python
#Train decision tree
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)
y_pred_dt = dt.predict(X_test)
acc_dt = accuracy_score(y_test, y_pred_dt)
```

```python
# train random forest
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)
y_pred_rf = rf.predict(X_test)
acc_rf = accuracy_score(y_test, y_pred_rf)
```

```python
print("Decision Tree Accuracy:", acc_dt)
print("Random Forest Accuracy:", acc_rf)
```
