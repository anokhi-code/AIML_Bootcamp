**Mean and Median of their respective heights, weights and Games played**

```python
import numpy as np

#Teams Data,
marvel = [
    {"S.no": 1, "Name": "IronMan", "Height(cm)": 182, "Weight(kg)": 90, "Games Played": 105},
    {"S.no": 2, "Name": "Thor", "Height(cm)": 187, "Weight(kg)": 120, "Games Played": 75},
    {"S.no": 3, "Name": "Captain America", "Height(cm)": 184, "Weight(kg)": 85, "Games Played": 205},
    {"S.no": 4, "Name": "SpiderMan", "Height(cm)": 175, "Weight(kg)": 75, "Games Played": 45},
    {"S.no": 5, "Name": "Hulk", "Height(cm)": 179, "Weight(kg)": 290, "Games Played": 210} 
]

dc = [
    {"S.no": 1, "Name": "Batman", "Height(cm)": 180, "Weight(kg)": 85, "Games Played": 105},
    {"S.no": 2, "Name": "Superman", "Height(cm)": 189, "Weight(kg)": 95, "Games Played": 300},
    {"S.no": 3, "Name": "Harvedent", "Height(cm)": 181, "Weight(kg)": 75, "Games Played": 55},
    {"S.no": 4, "Name": "Henery", "Height(cm)": 176, "Weight(kg)": 87, "Games Played": 125},
    {"S.no": 5, "Name": "Heralt", "Height(cm)": 184, "Weight(kg)": 100, "Games Played": 145}
]

def extract_feature(data,key):
    return np.array([item[key] for item in data])

for team_name, team in [("Marvel",marvel),("DC",dc)]:
    Heights = extract_feature(team,"Height(cm)")
    Weights = extract_feature(team,"Weight(kg)")
    GamesPlayed = extract_feature(team,"Games Played")
    print(f"\n{team_name} Team:")
    print(f"Mean Height: {np.mean(Heights)}, Median Height: {np.median(Heights)}")
    print(f"Mean Weight: {np.mean(Weights)}, Median Weight: {np.median(Weights)}")
    print(f"Mean Games Played: {np.mean(GamesPlayed)}, Median Games Played: {np.median(GamesPlayed)}")
```

**Deviation and Standard deviation of height and weight.**

```python
from scipy.stats import linregress

for team_name, team in [("Marvel", marvel),("DC", dc)]:
    Heights = extract_feature(team, "Height(cm)")
    Weights = extract_feature(team, "Weight(kg)")
    print(f"\n{team_name} Team:")
    print(f"Height - Deviation(Variance): {np.var(Heights, ddof=1):.2f}, Standard Deviation: {np.std(Heights, ddof=1):.2f}")
    print(f"Weight - Deviation(Variance): {np.var(Weights, ddof=1):.2f}, Standard Deviation: {np.std(Weights, ddof=1):.2f}")
```

**Doing Linear Regression (Relationship Between Weight and Games Played)**

```python
from scipy.stats import linregress

def linear_regression(team, team_name):
    Weights = extract_feature(team,"Weight(kg)")
    GamesPlayed = extract_feature(team,"Games Played")
    print(f"\nRegression for {team_name}:")
    result = linregress(Weights, GamesPlayed)
    print(f"y (Games Played) = {result.slope:.2f} * x (Weight(kg)) + {result.intercept:.2f}")
    print(f"correlation coefficient r = {result.rvalue:.2f}")
    print(f"p-value = {result.pvalue:.4f}, Standard error = {result.stderr:.2f}")

# Outside the function definition:
linear_regression(marvel, "Marvel")
linear_regression(dc, "DC")
```
